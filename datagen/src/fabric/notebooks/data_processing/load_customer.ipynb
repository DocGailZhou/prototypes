# Customer Data Loading

This notebook loads all customer-related CSV files from the lakehouse Files/data/customer directory into Delta tables.

## Tables to Load:
1. **Customer** - Master customer data with contact information and demographics
2. **CustomerAccount** - Customer account relationships and currency information
3. **CustomerRelationshipType** - Customer tier definitions and descriptions
4. **CustomerTradeName** - Business trade names and operating periods
5. **Location** - Customer addresses and geographic information

```python
# Setup and Configuration
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

# Schema Configuration
SCHEMA_NAME = "customer"
DATA_PATH = "Files/data/customer"

# Ensure schema exists
spark.sql(f"CREATE DATABASE IF NOT EXISTS {SCHEMA_NAME}")
print(f"✅ Schema '{SCHEMA_NAME}' ready!")
print(f"📁 Loading data from: {DATA_PATH}")
```

```python
# 1. Load Customer Table
print("👥 Loading Customer table...")

# Read CSV file
customer_df = spark.read.csv(
    f"{DATA_PATH}/Customer_Samples.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {customer_df.count()}")

# Write to Delta table
customer_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.Customer")

print(f"✅ Customer table loaded successfully!")
```

```python
# 2. Load CustomerAccount Table
print("💳 Loading CustomerAccount table...")

# Read CSV file
customer_account_df = spark.read.csv(
    f"{DATA_PATH}/CustomerAccount_Samples.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {customer_account_df.count()}")

# Write to Delta table
customer_account_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.CustomerAccount")

print(f"✅ CustomerAccount table loaded successfully!")
```

```python
# Summary and Verification
print("🎉 All customer tables loaded successfully!")
print("\n📅 Summary:")

# Show table counts
tables = ['Customer', 'CustomerAccount', 'CustomerRelationshipType', 'CustomerTradeName', 'Location']

for table in tables:
    count = spark.sql(f"SELECT COUNT(*) as count FROM {SCHEMA_NAME}.{table}").collect()[0]['count']
    print(f"   • {SCHEMA_NAME}.{table}: {count:,} records")

print(f"\n📁 All tables are available in the '{SCHEMA_NAME}' schema")
print("🚀 Ready for analytics and reporting!")
```
