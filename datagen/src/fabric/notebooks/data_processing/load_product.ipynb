# Product Data Loading

This notebook loads all product-related CSV files from the lakehouse Files/data/product directory into Delta tables.

## Tables to Load:
1. **Product** - Master product catalog with pricing, specifications, and lifecycle information
2. **ProductCategory** - Product category hierarchy with brand information and status

```python
# Setup and Configuration
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

# Schema Configuration
SCHEMA_NAME = "product"
DATA_PATH = "Files/data/product"

# Ensure schema exists
spark.sql(f"CREATE DATABASE IF NOT EXISTS {SCHEMA_NAME}")
print(f"✅ Schema '{SCHEMA_NAME}' ready!")
print(f"📁 Loading data from: {DATA_PATH}")
```

```python
# 1. Load Product Table
print("🛍️ Loading Product table...")

# Read CSV file
product_df = spark.read.csv(
    f"{DATA_PATH}/Product_Samples_Combined.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {product_df.count()}")

# Write to Delta table
product_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.Product")

print(f"✅ Product table loaded successfully!")
```

```python
# 2. Load ProductCategory Table
print("📂 Loading ProductCategory table...")

# Read CSV file
product_category_df = spark.read.csv(
    f"{DATA_PATH}/ProductCategory_Samples_Combined.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {product_category_df.count()}")

# Write to Delta table
product_category_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.ProductCategory")

print(f"✅ ProductCategory table loaded successfully!")
```

```python
# Summary and Verification
print("🎉 All product tables loaded successfully!")
print("\n📅 Summary:")

# Show table counts
tables = ['Product', 'ProductCategory']

for table in tables:
    count = spark.sql(f"SELECT COUNT(*) as count FROM {SCHEMA_NAME}.{table}").collect()[0]['count']
    print(f"   • {SCHEMA_NAME}.{table}: {count:,} records")

print(f"\n📁 All tables are available in the '{SCHEMA_NAME}' schema")
print("🚀 Ready for analytics and reporting!")
```
