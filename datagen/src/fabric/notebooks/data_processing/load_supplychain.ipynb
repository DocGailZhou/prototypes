# Supply Chain Data Loading

This notebook loads all supply chain-related CSV files from the lakehouse Files/data/supplychain directory into Delta tables.

## Tables to Load:
1. **Suppliers** - Master supplier data with backup relationships and performance metrics
2. **ProductSuppliers** - Product-supplier mappings with pricing and order parameters
3. **SupplyChainEvents** - Supply chain disruption events and impact tracking
4. **Warehouses** - Distribution centers and storage facilities with operational details

```python
# Setup and Configuration
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

# Schema Configuration
SCHEMA_NAME = "supplychain"
DATA_PATH = "Files/data/supplychain"

# Ensure schema exists
spark.sql(f"CREATE DATABASE IF NOT EXISTS {SCHEMA_NAME}")
print(f"✅ Schema '{SCHEMA_NAME}' ready!")
print(f"📁 Loading data from: {DATA_PATH}")
```

```python
# 1. Load Suppliers Table
print("🏭 Loading Suppliers table...")

# Read CSV file
suppliers_df = spark.read.csv(
    f"{DATA_PATH}/Suppliers.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {suppliers_df.count()}")

# Write to Delta table
suppliers_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.Suppliers")

print(f"✅ Suppliers table loaded successfully!")
```

```python
# 2. Load ProductSuppliers Table
print("🔗 Loading ProductSuppliers table...")

# Read CSV file
product_suppliers_df = spark.read.csv(
    f"{DATA_PATH}/ProductSuppliers.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {product_suppliers_df.count()}")

# Write to Delta table
product_suppliers_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.ProductSuppliers")

print(f"✅ ProductSuppliers table loaded successfully!")
```

```python
# 3. Load SupplyChainEvents Table
print("⚠️ Loading SupplyChainEvents table...")

# Read CSV file
supply_chain_events_df = spark.read.csv(
    f"{DATA_PATH}/SupplyChainEvents.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {supply_chain_events_df.count()}")

# Write to Delta table
supply_chain_events_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.SupplyChainEvents")

print(f"✅ SupplyChainEvents table loaded successfully!")
```

```python
# 4. Load Warehouses Table
print("🏭 Loading Warehouses table...")

# Read CSV file from inventory directory (warehouses are part of supply chain)
warehouses_df = spark.read.csv(
    "Files/data/inventory/Warehouses.csv",
    header=True,
    inferSchema=True
)

# Show record count
print(f"Records loaded: {warehouses_df.count()}")

# Write to Delta table in supplychain schema
warehouses_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.Warehouses")

print(f"✅ Warehouses table loaded successfully!")
```

```python
# Summary and Verification
print("🎉 All supply chain tables loaded successfully!")
print("\n📅 Summary:")

# Show table counts
tables = ['Suppliers', 'ProductSuppliers', 'SupplyChainEvents', 'Warehouses']

for table in tables:
    count = spark.sql(f"SELECT COUNT(*) as count FROM {SCHEMA_NAME}.{table}").collect()[0]['count']
    print(f"   • {SCHEMA_NAME}.{table}: {count:,} records")

print(f"\n📁 All tables are available in the '{SCHEMA_NAME}' schema")
print("🚀 Ready for analytics and reporting!")
```
