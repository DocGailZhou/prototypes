# Inventory Data Loading

This notebook loads all inventory-related CSV files from the lakehouse Files/data/inventory directory into Delta tables.

## Tables to Load:
1. **Inventory** - Current stock levels and warehouse information
2. **InventoryTransactions** - All stock movements and adjustments
3. **PurchaseOrders** - Purchase order headers with supplier information
4. **PurchaseOrderItems** - Purchase order line items with product details
5. **DemandForecast** - Predictive analytics for future demand planning

```python
# Setup and Configuration
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

# Schema Configuration
SCHEMA_NAME = "inventory"
DATA_PATH = "Files/data/inventory"

# Ensure schema exists
spark.sql(f"CREATE DATABASE IF NOT EXISTS {SCHEMA_NAME}")
print(f"✅ Schema '{SCHEMA_NAME}' ready!")
print(f"📁 Loading data from: {DATA_PATH}")
```

```python
# 1. Load Inventory Table
print("📦 Loading Inventory table...")

# Read CSV file
inventory_df = spark.read.csv(
    f"{DATA_PATH}/Inventory.csv",
    header=True,
    inferSchema=True
)

# Show sample data
print(f"Records loaded: {inventory_df.count()}")
print("Sample data:")
inventory_df.show(5, truncate=False)

# Write to Delta table
inventory_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.Inventory")

print(f"✅ Inventory table loaded successfully!")
```

```python
# 2. Load InventoryTransactions Table
print("🔄 Loading InventoryTransactions table...")

# Read CSV file
inventory_transactions_df = spark.read.csv(
    f"{DATA_PATH}/InventoryTransactions.csv",
    header=True,
    inferSchema=True
)

# Show sample data
print(f"Records loaded: {inventory_transactions_df.count()}")
print("Sample data:")
inventory_transactions_df.show(5, truncate=False)

# Write to Delta table
inventory_transactions_df.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable(f"{SCHEMA_NAME}.InventoryTransactions")

print(f"✅ InventoryTransactions table loaded successfully!")
```

```python
# Summary and Verification
print("🎉 All inventory tables loaded successfully!")
print("\n📅 Summary:")

# Show table counts
tables = ['Inventory', 'InventoryTransactions', 'PurchaseOrders', 'PurchaseOrderItems', 'DemandForecast']

for table in tables:
    count = spark.sql(f"SELECT COUNT(*) as count FROM {SCHEMA_NAME}.{table}").collect()[0]['count']
    print(f"   • {SCHEMA_NAME}.{table}: {count:,} records")

print(f"\n📁 All tables are available in the '{SCHEMA_NAME}' schema")
print("🚀 Ready for analytics and reporting!")
```
